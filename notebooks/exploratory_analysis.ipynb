Analysis of the Research Hub repository collection — datasets, citations, geographic coverage, and multilingual access metrics.

````python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns, warnings
warnings.filterwarnings('ignore'); sns.set_theme(style='whitegrid'); print('Libraries loaded')
````

````python
# Repository inventory
repos = pd.DataFrame({
    'repository': ['alzheimer-latinamerica-dashboard','mental-health-specialists-iberoamerica','human-brain-connectomics','generative-ai-academic-research','latamdata-py','research-hub'],
    'domain': ['Neurology','Mental Health','Neuroscience','AI/Education','Data Science','Multi-domain'],
    'languages': [8, 12, 6, 10, 4, 20],
    'has_doi': [True, True, True, True, True, True],
    'has_notebook': [True, True, True, True, False, True],
    'fair_score': [92, 89, 88, 85, 90, 95]
})
print(repos.to_string(index=False))
````

````python
fig,axes=plt.subplots(1,2,figsize=(13,5))
repos.plot.bar(x='repository',y='fair_score',ax=axes[0],legend=False,color='steelblue')
axes[0].set_title('FAIR Score by Repository'); axes[0].tick_params(axis='x',rotation=45); axes[0].set_ylim(0,100)
axes[0].axhline(repos['fair_score'].mean(),color='red',linestyle='--',label=f'Avg: {repos["fair_score"].mean():.0f}')
axes[0].legend()
repos.plot.bar(x='repository',y='languages',ax=axes[1],legend=False,color='green')
axes[1].set_title('Language Versions per Repository'); axes[1].tick_params(axis='x',rotation=45)
plt.tight_layout(); plt.show()
````
